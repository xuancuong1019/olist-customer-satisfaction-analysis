### Import dataset from olist_order_reviews_dataset.csv to staging.olist_order_reviews table

```python
import os
import pandas as pd
import gc
from dotenv import load_dotenv
from sqlalchemy import create_engine
import urllib

load_dotenv()  # đọc file .env

server = os.getenv('DB_SERVER')
database = os.getenv('DB_DATABASE')
driver = os.getenv('DB_DRIVER')

conn_str = urllib.parse.quote_plus(
    f'DRIVER={{{driver}}};SERVER={server};DATABASE={database};Trusted_Connection=yes;'
)
engine = create_engine(f'mssql+pyodbc:///?odbc_connect={conn_str}')

# Chỉ định định danh riêng file reviews
file_name = 'olist_order_reviews_dataset.csv'
table_name = 'olist_order_reviews'
path = rf'D:\olist-customer-satisfaction-analysis\data\raw\{file_name}'

print(f'Đang xử lý độc lập: {table_name}...')

try:
    chunk_size = 5000
    total_rows = 0
    
    for chunk in pd.read_csv(path, chunksize=chunk_size):
        
        # --- TIỀN XỬ LÝ LỖI NGẮT DÒNG ---
        # Thay thế các ký tự xuống dòng (\n, \r) bằng khoảng trắng để SQL không bị ngắt cột sai
        if 'review_comment_message' in chunk.columns:
            chunk['review_comment_message'] = chunk['review_comment_message'].str.replace(r'[\n\r]+', ' ', regex=True)
        if 'review_comment_title' in chunk.columns:
            chunk['review_comment_title'] = chunk['review_comment_title'].str.replace(r'[\n\r]+', ' ', regex=True)
        
        # Import vào bảng (Lưu ý: Nếu bảng cũ đang bị thiếu dòng, hãy đổi if_exists='append' thành 'replace' để ghi đè làm lại, hoặc TRUNCATE bảng trước trong SQL)
        chunk.to_sql(table_name, con=engine, schema='staging', if_exists='append', index=False)
        
        total_rows += len(chunk)
        del chunk
        gc.collect()
        
    print(f'✅ {table_name}: {total_rows} dòng đã nạp thành công.')
    
except Exception as e:
    print(f'❌ {table_name}: LỖI — {e}')

gc.collect() 
print("Quá trình Import độc lập kết thúc!")
```

### Import dataset from product_category_name_translation.csv to staging.product_category_name_translation table

```python
import os
import urllib
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine

# Tải biến môi trường từ file .env
load_dotenv()

# ==========================================
# 2. KHỞI TẠO KẾT NỐI DATABASE (DB CONNECTION)
# ==========================================
db_server = os.getenv('DB_SERVER')
db_database = os.getenv('DB_DATABASE')
db_driver = os.getenv('DB_DRIVER')

# Mã hóa chuỗi kết nối an toàn để tránh lỗi ký tự đặc biệt
connection_string = urllib.parse.quote_plus(
    f'DRIVER={{{db_driver}}};SERVER={db_server};DATABASE={db_database};Trusted_Connection=yes;'
)
db_engine = create_engine(f'mssql+pyodbc:///?odbc_connect={connection_string}')

# ==========================================
# 3. THỰC THI NẠP DỮ LIỆU (DATA IMPORT)
# ==========================================
# Khai báo đường dẫn và thông tin bảng đích
file_path = r'D:\olist-customer-satisfaction-analysis\data\raw\product_category_name_translation.csv'
target_table = 'product_category_name_translation'
target_schema = 'staging'

print(f"Đang xử lý nạp dữ liệu vào bảng: {target_schema}.{target_table}...")

try:
    # Do file translation rất nhỏ, đọc trực tiếp 1 lần bằng Pandas thay vì dùng chunk
    df_category_translation = pd.read_csv(file_path)
    
    # Đẩy dữ liệu thẳng vào SQL Server
    df_category_translation.to_sql(
        name=target_table, 
        con=db_engine, 
        schema=target_schema, 
        if_exists='append', 
        index=False
    )
    
    total_inserted = len(df_category_translation)
    print(f"✅ THÀNH CÔNG: Đã nạp {total_inserted} dòng vào bảng {target_schema}.{target_table}.")

except Exception as error:
    print(f"❌ THẤT BẠI khi nạp dữ liệu. Chi tiết lỗi: {error}")
```
